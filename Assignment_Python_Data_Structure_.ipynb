8. What is a hash table, and how does it relate to dictionaries in Python?
- A hash table is a data structure that stores key-value pairs in an array using a hash function to map keys to indices of the array.

How Hash Tables Work
(i). Key Hashing: When a key-value pair is added to a hash table, the key is hashed to generate a unique index.
(ii). Indexing: The hashed key is used to index into the array, where the corresponding value is stored.
(ii). Collision Resolution: If two keys hash to the same index, a collision resolution mechanism is used to handle the conflict.

Relation to Dictionaries in Python
(i). Implementation: Dictionaries in Python are implemented using hash tables.
(ii). Fast Lookups: Hash tables enable fast lookups, insertions, and deletions, making dictionaries efficient for storing and retrieving data.
(iii). Key-Value Pairs: Dictionaries store key-value pairs, which is similar to the key-value pairs stored in hash tables.

Benefits of Hash Tables in Dictionaries
(i). Fast Lookup: Hash tables provide fast lookup operations, with an average time complexity of O(1).
(ii). Efficient Insertion: Hash tables enable efficient insertion of new key-value pairs.
(iii). Flexible Data Structure: Dictionaries can store a wide range of data types, including strings, integers, floats, and other dictionaries.

17. Describe the time complexity of accessing elements in a dictionary?
- The time complexity of accessing elements in a dictionary is typically O(1), making it very efficient.

Why Dictionaries are Fast
(i). Hash Table Implementation: Dictionaries are implemented using hash tables, which enable fast lookups.
(ii). Average Case: In the average case, dictionary lookups take constant time, O(1), regardless of the size of the dictionary.

Worst-Case Scenario
(i). Hash Collisions: In the worst-case scenario, where multiple keys hash to the same index (known as hash collisions), the time complexity can degrade to O(n), where n is the number of keys in the dictionary.
(ii). Rare Occurrence: However, hash collisions are rare in practice, and Python's dictionary implementation is designed to minimize their occurrence.

Benefits of Fast Lookups
(i). Efficient Data Retrieval: Fast lookups make dictionaries ideal for applications where data needs to be retrieved quickly.
(ii). Scalability: Dictionaries can handle large amounts of data without significant performance degradation.
